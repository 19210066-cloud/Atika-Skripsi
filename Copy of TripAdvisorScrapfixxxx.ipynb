```python
#fixxxxscrapreviewhotelsulut

import urllib.parse
import requests
import pandas as pd
from google.colab import files

# API KEY SERPAPI
API_KEY = "8bee44f8c9bae3f132961558421d168edef68fe7bd8f455ccd14d4ea1555dde6"

# Kata kunci untuk mencari review hotel di Sulawesi Utara
keywords = [
    "review hotel manado",
    "review hotel sulawesi utara",
    "pengalaman menginap di hotel sulawesi utara"
]

# List untuk menyimpan data
all_url = []
all_title = []
all_snippet = []
all_source = []
all_keyword = []

# Loop setiap keyword
for string in keywords:

    print("Mengambil data untuk keyword:", string)

    # Pagination Google
    for i in range(0, 500, 10):

        params = {
            'q': string,
            'location': "Indonesia",
            'hl': "id",
            'gl': "id",
            'api_key': "8bee44f8c9bae3f132961558421d168edef68fe7bd8f455ccd14d4ea1555dde6",
            'start': i
        }

        # Encode parameter
        param = urllib.parse.urlencode(params)

        # Request ke SERPAPI
        r = requests.get('https://serpapi.com/search.json?' + param)

        # Cek status request
        if r.status_code != 200:
            print("Request error:", r.status_code)
            break

        data = r.json()

        # Cek jika API error
        if "error" in data:
            print("API Error:", data["error"])
            break

        # Ambil hasil pencarian
        if 'organic_results' in data:

            for item in data['organic_results']:

                all_url.append(item.get('link', ''))
                all_title.append(item.get('title', ''))
                all_snippet.append(item.get('snippet', ''))
                all_source.append(item.get('displayed_link', ''))
                all_keyword.append(string)

# Membuat DataFrame
df = pd.DataFrame({
    "keyword": all_keyword,
    "url": all_url,
    "title": all_title,
    "snippet": all_snippet,
    "source": all_source
})

# Menghapus duplikasi
df = df.drop_duplicates(subset="url")

# Nama file
filename = "Dataset_Review_Hotel_Sulut.xlsx"

# Simpan ke Excel
df.to_excel(filename, index=False)

print("Jumlah data yang berhasil dikumpulkan:", len(df))

# Download file
files.download(filename)
```
